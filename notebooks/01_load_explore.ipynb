## 3. Reservoir Sampling (Memory-Safe Random Inspection)

```python
samples = sample_lines(LOG_PATH, k=5, seed=42)
for idx, sample in enumerate(samples, 1):
    print(f'Sample {idx}: {sample}')
```

## 5. Security Token Value Analysis

| Token | Example Value | Cyber-Security Value |
| :--- | :--- | :--- |
| **IP Address** | `198.51.100.7` | Attacker identity, geo-location, burst volumes, reputation |
| **Timestamp** | `[10/Oct/2026:13:55:36 +0000]` | Time-of-day profiling, brute-force frequency windows |
| **HTTP Method** | `POST` / `GET` | POST to `/login` hints authentication attacks; unusual verbs (`PUT`, `DELETE`, `DEBUG`) |
| **URL / Query** | `/products.php?id=1%27+OR+1=1` | Direct injection payloads (SQLi, XSS, path traversal, scans) |
| **Status Code** | `404`, `401`, `500` | Scanners trigger 404 clusters; brute-force generates 401; exploit crashes trigger 500 |
| **Bytes Sent** | `0` or `500000` | Outliers indicate data exfiltration or blocked requests |
| **User-Agent** | `sqlmap/1.7`, `Nikto` | Automated penetration tools vs real browser signatures |
