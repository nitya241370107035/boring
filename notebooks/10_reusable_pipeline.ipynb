### Objective:
Package log ingestion, parsing, cleaning, rule-based labeling, and feature engineering into an end-to-end, reusable command-line interface (CLI) data pipeline with structured logging, execution auditing (`run_summary.json`), and streaming tailing support.

## 1. Imports and Setup

```python
import sys
from pathlib import Path
import json
import pandas as pd
sys.path.append('..')

from pipeline.run import run_pipeline

LOG_PATH = Path('../tests/fixtures/sample_access.log')
RULES_PATH = Path('../rules/web_rules.yaml')
OUT_PARQUET = Path('../data/processed/pipeline_out.parquet')
OUT_SUMMARY = Path('../run_summary.json')
```

## 2. Execute End-to-End Pipeline

```python
df, summary = run_pipeline(
    logfile=LOG_PATH,
    rules_path=RULES_PATH,
    out_path=OUT_PARQUET,
    summary_path=OUT_SUMMARY
)
print(f'Pipeline execution complete in {summary["duration_seconds"]} seconds.')
print(f'Total processed rows: {len(df)}')
```

## 3. Verify Execution Summary Audit JSON

```python
with open(OUT_SUMMARY, 'r', encoding='utf-8') as f:
    audit_data = json.load(f)

for k, v in audit_data.items():
    print(f'{k}: {v}')
```

## 4. Parquet Lake Verification

```python
parquet_df = pd.read_parquet(OUT_PARQUET)
print(f'Parquet schema verified. Columns: {len(parquet_df.columns)}, Rows: {len(parquet_df)}')
parquet_df[['ip', 'method', 'path', 'label', 'url_entropy', 'win_req_count']].head(8)
```
